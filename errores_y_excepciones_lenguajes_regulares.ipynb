```python
usuarios = {
    "ANA": "1234",
    "JUAN": "acd"
}
try:
  usuario = input("Usuario: ").upper()
  password = input("Contraseña: ")

  if usuario not in usuarios:
    raise KeyError("El usuario no existe")
  if usuarios[usuario] != password:
    raise ValueError("Contraseña incorrecta")
  else:
    print("Bienvenido, usuario")

except KeyError as e:
  print("Error:", e)
except ValueError as e:
  print("Error:", e)
```

```python
try:
  num1 = float(input("Número 1: "))
  num2 = float(input("Número 2: "))
  op = input("Operación (+, -, *, /, %): ")

  if op == "+":
    resultado = num1 + num2
  elif op == "-":
    resultado = num1 - num2
  elif op == "*":
    resultado = num1 * num2
  elif op == "/":
    resultado = num1 / num2
  elif op == "%":
    resultado = num1 % num2
  else:
    raise ValueError("Operación no válida")

  print("Resultado:", resultado)

except ValueError as e:
  print("Error de valor:", e)
except ZeroDivisionError:
  print("No se puede dividir entre cero", e)
except TypeError as e:
  print("Error de tipo:", e)
```

```python
try:
  productos = {
      "pan": 1.5,
      "leche": 2.0,
      "huevos": 3.0
  }

  print("Productos disponibles:", productos)

  nombre = input("¿Qué producto quieres?: ").lower()
  cantidad = int(input("Cantidad: "))

  if cantidad <= 0:
    raise ValueError("La cantidad debe ser mayor a 0")

  total = productos[nombre] * cantidad
  print("Total:", total)
except ValueError as e:
  print("Error de valor:", e)
except KeyError as e:
  print("Producto no existente:", e)
```

```python
try:

    datos = input("Ingresa notas separadas por coma: ")
    notas = [float(n) for n in datos.split(",")]
    if len(notas) == 0:
        raise ValueError("No ingresaste notas")
    promedio = sum(notas) / len(notas)
    posicion = int(input("¿Qué nota quieres ver (índice)?: "))
    if posicion < 0:
        raise ValueError("Ingresa solo índices positivos")
    print("Nota seleccionada:", notas[posicion])
    print("Promedio:", promedio)

except ValueError as e:
    print("Error de valor:", e)

except ZeroDivisionError:
    print("No se puede dividir entre cero")

except IndexError:
    print("Índice fuera de rango")
```

```python
import re
```

```python
texto = "Python es un petuche"
resultado = re.search("Python", texto)
print(resultado)
```

```python
resultado1 = re.search("x", texto)
print(resultado1)
```

```python
texto = "Tengo 201 perros y 3 gatos"
resultado3 = re.findall(r"\d", texto)
print(resultado3)
```

El método findall permite encontrar números dentro de cadenas

```python
texto = "Python es divertido"
resultado4 = re.findall(r"\w", texto)
print(resultado4)
```

```python
texto = "Python es divertido"
resultado4 = re.findall(r"\w+", texto)
print(resultado4)
```
